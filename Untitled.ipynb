```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.cluster import KMeans
from sklearn.preprocessing import StandardScaler

# Cấu hình hiển thị biểu đồ
plt.rcParams['font.family'] = 'DejaVu Sans'
plt.rcParams['figure.figsize'] = (12, 8)

file_excel_goc = 'data_goc.xlsx' 
file_csv_sach = 'du_lieu_7_mon_cot_loi.csv'

# DANH SÁCH 7 MÔN QUAN TRỌNG BẠN CHỌN
cac_mon_chon = [
    'Tin học trong kỹ thuật', 
    'Toán rời rạc',
    'Cấu trúc dữ liệu và giải thuật',
    'Cơ sở dữ liệu',
    'Phân tích và thiết kế hệ thống', 
    'Kỹ thuật lập trình nâng cao',
    'Đồ án mạng máy tính'
]
```

```python
# 1. Đọc file Excel thô
df_raw = pd.read_excel(file_excel_goc)

# 2. Thu thập thông tin Sinh viên
mssv_list = df_raw.iloc[0, 3:].values
ten_sv_list = df_raw.iloc[1, 3:].values
danh_sach_sv = [f"{mssv}_{ten}" for mssv, ten in zip(mssv_list, ten_sv_list)]

# 3. Lấy ma trận điểm và gán tên môn học
df_diem_raw = df_raw.iloc[2:, :]
df_diem_raw.columns = ['STT', 'Ma_Mon', 'Ten_Mon'] + list(df_raw.columns[3:])

# LỌC: Chỉ giữ lại dòng của 7 môn học quan trọng
df_7_mon = df_diem_raw[df_diem_raw['Ten_Mon'].isin(cac_mon_chon)].copy()

# 4. Làm sạch ký tự lạ
matrix_diem = df_7_mon.iloc[:, 3:].copy()
matrix_diem = matrix_diem.astype(str).replace(['x', '-', '—', '\|', 'nan', 'NaN', 'Nan', 'nAn', '/'], np.nan, regex=True)

# Đổi dấu phẩy thành dấu chấm
for col in matrix_diem.columns:
    matrix_diem[col] = matrix_diem[col].str.replace(',', '.').astype(float)

# 5. Chuyển vị ma trận (Sinh viên thành Dòng, 7 Môn học thành Cột)
df_csv = matrix_diem.T
df_csv.index = danh_sach_sv
df_csv.columns = df_7_mon['Ten_Mon'].values

# 6. Xuất ra file CSV sạch chỉ chứa 7 môn cốt lõi
df_csv.to_csv(file_csv_sach, index_label='Sinh_Vien')
print(f"--> Đã lọc và xuất file CSV chứa 7 môn cốt lõi tại: {file_csv_sach}")
print(f"Kích thước ma trận mới: {df_csv.shape} (71 Sinh viên, 7 Môn học)")
```

```python
# 1. Đọc file CSV 7 môn
df_Kmeans = pd.read_csv(file_csv_sach, index_col='Sinh_Vien')

# 2. Xử lý ô trống (NaN) bằng điểm trung bình môn học đó
df_imputed = df_Kmeans.fillna(df_Kmeans.mean())
df_imputed = df_imputed.fillna(0) # Phòng hờ

# 3. Chuẩn hóa thang đo về Mean=0, Std=1 để chạy K-means chính xác
scaler = StandardScaler()
X_scaled = scaler.fit_transform(df_imputed)

print("Dữ liệu 7 môn chuyên ngành đã sẵn sàng!")
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.cluster import KMeans
from sklearn.manifold import TSNE

# 1. Chạy K-means với K=3 trên không gian 7 môn học
kmeans = KMeans(n_clusters=3, random_state=42, n_init=10)
labels = kmeans.fit_predict(X_scaled)

# Lấy tọa độ tâm cụm trong không gian 7 chiều gốc
centroids_7d = kmeans.cluster_centers_

# 2. Đồng bộ hóa dữ liệu Sinh viên và Tâm cụm để chạy t-SNE cùng lúc
# Việc này giúp các Tâm cụm được ánh xạ chuẩn xác vào giữa các nhóm chấm sinh viên
X_combined = np.vstack([X_scaled, centroids_7d])

tsne = TSNE(n_components=2, perplexity=15, random_state=42, max_iter=1000)
X_tsne_combined = tsne.fit_transform(X_combined)

# Tách tọa độ t-SNE của sinh viên và tâm cụm ra riêng sau khi tính toán xong
X_tsne_sv = X_tsne_combined[:-3, :]
X_tsne_centroids = X_tsne_combined[-3:, :]

# Tạo DataFrame vẽ đồ thị chấm sinh viên
df_visual = pd.DataFrame(data=X_tsne_sv, columns=['Tọa độ t-SNE X', 'Tọa độ t-SNE Y'])
df_visual['Sinh_Vien'] = df_imputed.index
df_visual['Nhóm'] = labels
df_visual['Nhóm'] = df_visual['Nhóm'].map({0: 'Nhóm 0', 1: 'Nhóm 1', 2: 'Nhóm 2'})

# 3. TRỰC QUAN HÓA: Vẽ đồ thị phân cụm chấm (Scatter Plot)
plt.figure(figsize=(12, 8))
mau_sac_cum = ['#d62728', '#1f77b4', '#2ca02c'] # Đỏ, Xanh dương, Xanh lá

# BƯỚC A: Vẽ 71 sinh viên thành các chấm tròn màu
sns.scatterplot(
    x='Tọa độ t-SNE X', 
    y='Tọa độ t-SNE Y', 
    hue='Nhóm', 
    palette=mau_sac_cum, 
    data=df_visual, 
    s=150, 
    alpha=0.8,
    edgecolor='black',
    linewidth=1
)

# BƯỚC B: VẼ CÁC TÂM CỤM (CENTROIDS) LÊN ĐỒ THỊ t-SNE
# Sử dụng ký hiệu 'X' màu vàng viền đen, kích thước rất to để làm nổi bật trọng tâm từng cụm
plt.scatter(
    X_tsne_centroids[:, 0], 
    X_tsne_centroids[:, 1], 
    marker='X', 
    s=350, 
    c='yellow', 
    edgecolor='black', 
    linewidth=2,
    label='Tâm cụm (Centroid)'
)

# Thêm mã số sinh viên (MSSV) thu nhỏ lên từng chấm tròn
for i in range(len(df_visual)):
    mssv_ngan = df_visual['Sinh_Vien'].iloc[i].split('_')[0]
    plt.text(
        df_visual['Tọa độ t-SNE X'].iloc[i] + 0.15, 
        df_visual['Tọa độ t-SNE Y'].iloc[i] + 0.15, 
        mssv_ngan, 
        fontsize=7, 
        alpha=0.6,
        fontweight='semibold'
    )

plt.title('ĐỒ THỊ PHÂN CỤM SINH VIÊN QUA LỌC 7 MÔN CỐT LÕI\n(Hiển thị rõ ràng 3 quần thể và vị trí Tâm cụm Centroid)', fontsize=14, fontweight='bold')
plt.grid(True, linestyle=':', alpha=0.6)
plt.legend(loc='best', fontsize=11)
plt.tight_layout()
plt.show()


# =========================================================================
# ĐỒ THỊ 2: BIỂU ĐỒ ĐƯỜNG ĐẶC TRƯNG NĂNG LỰC CỦA 3 NHÓM QUA 7 MÔN
# =========================================================================
plt.figure(figsize=(14, 6))
df_analysis = df_imputed.copy()
df_analysis['Nhóm'] = labels
df_centroids = df_analysis.groupby('Nhóm').mean()

for nhom_id in df_centroids.index:
    plt.plot(
        df_centroids.columns, 
        df_centroids.loc[nhom_id].values, 
        marker='o', 
        markersize=10,
        linewidth=3, 
        color=mau_sac_cum[nhom_id],
        label=f'Đặc trưng Nhóm {nhom_id}'
    )

plt.title('BIỂU ĐỒ ĐƯỜNG SO SÁNH NĂNG LỰC 3 NHÓM SINH VIÊN QUA 7 MÔN CỐT LÕI', fontsize=14, fontweight='bold')
plt.xlabel('Tên 7 Môn Học Chuyên Ngành', fontsize=12)
plt.ylabel('Điểm trung bình hệ 4 (0.0 - 4.0)', fontsize=12)
plt.ylim(0, 4.2)
plt.xticks(rotation=15, fontsize=10)
plt.grid(True, linestyle='--', alpha=0.6)
plt.legend(loc='lower left', fontsize=12)
plt.tight_layout()
plt.show()


# =========================================================================
# 4. IN THỐNG KÊ VÀ XUẤT FILE CSV KẾT QUẢ
# =========================================================================
print("\n" + "="*25 + " THỐNG KÊ PHÂN BỔ THÀNH VIÊN " + "="*25)
for nhom_id in sorted(df_analysis['Nhóm'].unique()):
    SVs = df_analysis[df_analysis['Nhóm'] == nhom_id].index.values
    print(f"\n➔ Nhóm {nhom_id} (Có {len(SVs)} sinh viên):")
    print(f"   Ví dụ thành viên: {', '.join(SVs[:5])} ...")
print("="*75)

df_analysis[['Nhóm']].to_csv('ket_qua_phan_cum_7_mon.csv', encoding='utf-8-sig')
print("\n--> Đã lưu danh sách phân nhóm hoàn chỉnh vào file 'ket_qua_phan_cum_7_mon.csv'!")
```
